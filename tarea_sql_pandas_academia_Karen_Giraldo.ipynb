## 3. Cargar tablas a SQLite

```python
conn = sqlite3.connect(":memory:")

estudiantes.to_sql("estudiantes", conn, index=False, if_exists="replace")
cursos.to_sql("cursos", conn, index=False, if_exists="replace")
inscripciones.to_sql("inscripciones", conn, index=False, if_exists="replace")
pagos.to_sql("pagos", conn, index=False, if_exists="replace")

def run_sql(query):
    return pd.read_sql_query(query, conn)
```

### Punto 4. Agregación simple

Calcula:
- número total de cursos
- precio promedio
- precio mínimo
- precio máximo

Usa la tabla `cursos`.

**Entregable:**
- SQL
- pandas

```python
# Punto 4 - SQL
query_4 = """
SELECT
    COUNT(*) AS total_cursos,
    AVG(precio) AS precio_promedio,
    MIN(precio) AS precio_minimo,
    MAX(precio) AS precio_maximo
FROM cursos;
"""

resumen_cursos = run_sql(query_4)
resumen_cursos
```

```python
# Punto 4 - pandas
resumen_cursos_pd = pd.DataFrame({
    "total_cursos": [cursos["curso_id"].count()],
    "precio_promedio": [cursos["precio"].mean()],
    "precio_minimo": [cursos["precio"].min()],
    "precio_maximo": [cursos["precio"].max()]
})

resumen_cursos_pd
```

### Punto 5. GROUP BY

Cuenta cuántos cursos hay por `categoria`.

Ordena de mayor a menor cantidad.

**Entregable:**
- SQL
- pandas

```python
# Punto 5 - SQL
query_5 = """
SELECT
    categoria,
    COUNT(*) AS cantidad_cursos
FROM cursos
GROUP BY categoria
ORDER BY cantidad_cursos DESC;
"""

cursos_por_categoria = run_sql(query_5)
cursos_por_categoria
```

```python
# Punto 5 - pandas
cursos_por_categoria_pd = (
    cursos.groupby("categoria")
    .size()
    .reset_index(name="cantidad_cursos")
    .sort_values(by="cantidad_cursos", ascending=False)
)

cursos_por_categoria_pd
```

### Punto 6. JOIN entre tablas

Muestra:
- nombre del estudiante
- nombre del curso
- estado de la inscripción

Usa `estudiantes`, `inscripciones` y `cursos`.

**Entregable:**
- SQL
- pandas

```python
# Punto 6 - SQL
query_6 = """
SELECT
    estudiantes.nombre,
    cursos.nombre_curso,
    inscripciones.estado
FROM estudiantes
JOIN inscripciones
    ON estudiantes.estudiante_id = inscripciones.estudiante_id
JOIN cursos
    ON inscripciones.curso_id = cursos.curso_id;
"""

estudiantes_cursos = run_sql(query_6)
estudiantes_cursos
```

```python
# Punto 6 - pandas
estudiantes_cursos_pd = (
    inscripciones
    .merge(estudiantes[["estudiante_id", "nombre"]], on="estudiante_id")
    .merge(cursos[["curso_id", "nombre_curso"]], on="curso_id")
    [["nombre", "nombre_curso", "estado"]]
)

estudiantes_cursos_pd
```

### Punto 7. LEFT JOIN

Trae todos los estudiantes, incluso si no tuvieran inscripciones, mostrando:
- `estudiante_id`
- `nombre`
- `inscripcion_id`

**Entregable:**
- SQL
- pandas

> Pista: en este dataset puede que todos tengan al menos una inscripción, pero igual debes practicar el `LEFT JOIN`.

```python
# Punto 7 - SQL
query_7 = """
SELECT
    estudiantes.estudiante_id,
    estudiantes.nombre,
    inscripciones.inscripcion_id
FROM estudiantes
LEFT JOIN inscripciones
    ON estudiantes.estudiante_id = inscripciones.estudiante_id;
"""

estudiantes_inscripciones = run_sql(query_7)
estudiantes_inscripciones
```

```python
# Punto 7 - pandas
estudiantes_inscripciones_pd = estudiantes.merge(
    inscripciones[["estudiante_id", "inscripcion_id"]],
    on="estudiante_id",
    how="left"
)

estudiantes_inscripciones_pd
```

### Punto 8. HAVING

Calcula cuántas inscripciones tiene cada estudiante y muestra solo aquellos con **2 o más**.

Debes mostrar:
- `estudiante_id`
- `nombre`
- `n_inscripciones`

**Entregable:**
- SQL
- pandas

```python
# Punto 8 - SQL
query_8 = """
SELECT
    estudiantes.estudiante_id,
    estudiantes.nombre,
    COUNT(inscripciones.inscripcion_id) AS n_inscripciones
FROM estudiantes
JOIN inscripciones
    ON estudiantes.estudiante_id = inscripciones.estudiante_id
GROUP BY estudiantes.estudiante_id, estudiantes.nombre
HAVING COUNT(inscripciones.inscripcion_id) >= 2;
"""

estudiantes_con_varias_inscripciones = run_sql(query_8)
estudiantes_con_varias_inscripciones
```

```python
# Punto 8 - pandas
estudiantes_con_varias_inscripciones_pd = (
    inscripciones
    .groupby("estudiante_id")
    .size()
    .reset_index(name="n_inscripciones")
    .query("n_inscripciones >= 2")
    .merge(estudiantes[["estudiante_id", "nombre"]], on="estudiante_id")
    [["estudiante_id", "nombre", "n_inscripciones"]]
)

estudiantes_con_varias_inscripciones_pd
```

### Punto 9. CASE WHEN / segmentación

Clasifica los cursos según su precio:
- `Bajo` si el precio es menor a 120
- `Medio` si el precio está entre 120 y 180 inclusive
- `Alto` si el precio es mayor a 180

Debes mostrar:
- `nombre_curso`
- `precio`
- `segmento_precio`

**Entregable:**
- SQL
- pandas

```python
# Punto 9 - SQL
query_9 = """
SELECT
    nombre_curso,
    precio,
    CASE
        WHEN precio < 120 THEN 'Bajo'
        WHEN precio BETWEEN 120 AND 180 THEN 'Medio'
        ELSE 'Alto'
    END AS segmento_precio
FROM cursos;
"""

cursos_segmentados = run_sql(query_9)
cursos_segmentados
```

```python
# Punto 9 - pandas
cursos_segmentados_pd = cursos[["nombre_curso", "precio"]].copy()

cursos_segmentados_pd["segmento_precio"] = cursos_segmentados_pd["precio"].apply(
    lambda precio: "Bajo" if precio < 120
    else "Medio" if precio <= 180
    else "Alto"
)

cursos_segmentados_pd
```

### Punto 10. Ejercicio final integrador

Calcula el total pagado por categoría de curso.

Debes:
- unir `cursos`, `inscripciones` y `pagos`
- agrupar por `categoria`
- sumar `monto`
- ordenar de mayor a menor

**Entregable:**
- SQL
- pandas

```python
# Punto 10 - SQL
query_10 = """
SELECT
    cursos.categoria,
    SUM(pagos.monto) AS total_pagado
FROM cursos
JOIN inscripciones
    ON cursos.curso_id = inscripciones.curso_id
JOIN pagos
    ON inscripciones.inscripcion_id = pagos.inscripcion_id
GROUP BY cursos.categoria
ORDER BY total_pagado DESC;
"""

total_pagado_categoria = run_sql(query_10)
total_pagado_categoria
```

```python
# Punto 10 - pandas
total_pagado_categoria_pd = (
    inscripciones
    .merge(cursos[["curso_id", "categoria"]], on="curso_id")
    .merge(pagos[["inscripcion_id", "monto"]], on="inscripcion_id")
    .groupby("categoria", as_index=False)["monto"]
    .sum()
    .rename(columns={"monto": "total_pagado"})
    .sort_values(by="total_pagado", ascending=False)
)

total_pagado_categoria_pd
```
